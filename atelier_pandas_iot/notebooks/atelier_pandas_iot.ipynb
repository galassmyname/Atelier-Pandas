## Partie 1 – Series

### 1) Series de 4 températures

```python
import pandas as pd
import numpy as np

temperatures = pd.Series([22.5, 23.1, 24.8, 21.9])
print(temperatures)
```

### 2) Afficher l'index

```python
print(list(temperatures.index))
```

### 3) Afficher la 1ère valeur

```python
print(temperatures[0])
```

### 4) Afficher la dernière valeur

```python
print(temperatures.iloc[-1])
```

### 5) Series de 4 températures avec index = heures (12h, 13h, 14h, 15h) — 2 façons

```python
# Façon 1 : en passant l'index directement à la création
temp_heures = pd.Series([22.1, 23.8, 24.8, 21.9], index=["12h", "13h", "14h", "15h"])
print(temp_heures)

# Façon 2 : en créant la Series puis en assignant l'index après coup
temp_heures2 = pd.Series([22.5, 23.1, 24.8, 21.9])
temp_heures2.index = ["12h", "13h", "14h", "15h"]
print(temp_heures2)
```

### 6) 1ère et dernière valeur

```python
print(temp_heures.iloc[0])
print(temp_heures.iloc[-1])
```
