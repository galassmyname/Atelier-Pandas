### 5) Series de 4 températures avec index = heures (12h, 13h, 14h, 15h) — 2 façons

```python
# Façon 1 : en passant l'index directement à la création
temp_heures = pd.Series([22.1, 23.8, 24.8, 21.9], index=["12h", "13h", "14h", "15h"])
print(temp_heures)

# Façon 2 : en créant la Series puis en assignant l'index après coup
temp_heures2 = pd.Series([22.5, 23.1, 24.8, 21.9])
temp_heures2.index = ["12h", "13h", "14h", "15h"]
print(temp_heures2)
```

### 6) 1ère et dernière valeur

```python
print(temp_heures.iloc[0])
print(temp_heures.iloc[-1])
```

## Partie 5 – Manipulation des colonnes

### 1) Ajouter la colonne "temperature_fahrenheit"

```python
df["temperature_fahrenheit"] = df["temperature"] * 9/5 + 32
print(df.head())
```

### 2) Ajouter la colonne "niveau_temperature" (Élevée si > 30, sinon Normale)

```python
df["niveau_temperature"] = np.where(df["temperature"] > 30, "Élevée", "Normale")
print(df[["temperature", "niveau_temperature"]].head())
```

### 3) Renommer la colonne "humidite" en "humidite_relative"

```python
df = df.rename(columns={"humidite": "humidite_relative"})
print(df.columns)
```

### 4) Supprimer la colonne "temperature_fahrenheit"

```python
df = df.drop(columns=["temperature_fahrenheit"])
print(df.columns)
```

## Partie 6 – Filtrage

### 1) Lignes telles que température > 30

```python
df_tempelevee = df[df["temperature"] > 30]
print(df_tempelevee)
```

### 2) Lignes telles que température > 30 et humidité > 70

```python
df_temphumidite = df[(df["temperature"] > 30) & (df["humidite_relative"] > 70)]
print(df_temphumidite)
```

## Partie 9 – Gestion des valeurs manquantes

### 1) Nombre de valeurs manquantes par colonne

```python
df.isnull().sum()
```

### 2) Nombre total de valeurs manquantes

```python
df.isnull().sum().sum()
```

### 3) Taux de valeurs manquantes

```python
taux_manquant = df.isnull().sum().sum() / df.size * 100
print(f"{taux_manquant:.2f} %")
```

### 4) Lignes contenant des valeurs manquantes

```python
df[df.isnull().any(axis=1)]
```

### 5) Température et humidité : remplacer les valeurs manquantes par la moyenne

```python
df["temperature"] = df["temperature"].fillna(df["temperature"].mean())
df["humidite_relative"] = df["humidite_relative"].fillna(df["humidite_relative"].mean())
print(df[["temperature", "humidite_relative"]].isnull().sum())
```

### 6) Consommation : remplacer les valeurs manquantes par la médiane

```python
df["consommation"] = df["consommation"].fillna(df["consommation"].median())
print(df["consommation"].isnull().sum())
```

### 7) "etat" : remplacer les valeurs manquantes par "INCONNU"

```python
df["etat"] = df["etat"].fillna("INCONNU")
print(df["etat"].isnull().sum())
```
